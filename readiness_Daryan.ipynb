```python
def calculate_mean_and_max(numbers):
    mean = sum(numbers) / len(numbers)
    maximum = max(numbers)
    return mean, maximum

my_list = [10, 20, 30, 40, 50]
mean, max_val = calculate_mean_and_max(my_list)
print(f"Mean: {mean}, Maximum: {max_val}")
```
